```python


import numpy as np
import pandas as pd
import re
from nltk.corpus import stopwords
from sklearn.model_selection import train_test_split
from nltk.stem.porter import PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score


```

```python
import nltk
nltk.download('stopwords')
```

```python
print(stopwords.words('english'))
```

```python
dataset = pd.read_csv('/content/archive.zip',encoding='ISO-8859-1')
```

```python
dataset.head()
```

```python
col_names = ['target', 'id', 'date', 'flag', 'user', 'text']
dataset.columns = col_names
dataset.head()
```

```python
dataset.shape
```

```python
dataset.isnull().sum()
```

```python
dataset['target'].value_counts()
```

```python
dataset['target'] = dataset['target'].map({4:1,0:0})
```

```python
stremmer = PorterStemmer()

def stemming(content):
  stemmed_content = re.sub('[^a-zA-Z]',' ',content)
  stemmed_content = stemmed_content.lower()
  stemmed_content = stemmed_content.split()
  stemmed_content = [stremmer.stem(word) for word in stemmed_content if word not in stopwords.words('english')]
  stemmed_content = ' '.join(stemmed_content)
  return stemmed_content
```

```python
dataset['text'] = dataset['text'].apply(stemming)
```

```python
dataset.head()
```

```python
x = dataset['text']
y = dataset['target']
```

```python
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2 , random_state= 0)
```

```python
vectorizer = TfidfVectorizer()
x_train = vectorizer.fit_transform(x_train)
x_test = vectorizer.transform(x_test)
```

```python
print(x_train)
```

```python
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
print(accuracy_score(y_test,y_pred))
```

```python
from sklearn.metrics import classification_report
y_pred = model.predict(x_test)
print(classification_report(y_test, y_pred))
```

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
svm_model = LinearSVC()
svm_model.fit(x_train, y_train)
```

```python
y_pred = svm_model.predict(x_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
```

```python
from sklearn.model_selection import train_test_split
import pandas as pd


x = dataset['text']
y = dataset['target']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
vectorizer = CountVectorizer()


x_train = vectorizer.fit_transform(x_train)

x_test = vectorizer.transform(x_test)

dt_model = DecisionTreeClassifier(criterion='entropy', max_depth=5, random_state=0)
dt_model.fit(x_train, y_train)
```

```python
y_pred = dt_model.predict(x_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
def predict_sentiment(text):
  text = re.sub('[^a-zA-Z]',' ',text)
  text = text.lower()
  text = text.split()
  text = [stremmer.stem(word) for word in text if not word in stopwords.words('english')]
  text = ' '.join(text)
  text = [text]
  text = vectorizer.transform(text)
  sentiment = model.predict(text)
  if sentiment == 0:
    return "Negative"
  else:
    return "Positive"
```

```python
print(predict_sentiment("I love this movie"))
print(predict_sentiment("I hate this movie"))
```
